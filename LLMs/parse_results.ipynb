```python
import pandas as pd

# prints confusion matrix, accuracy, and 5 random FP and FN
def parse_results(csv_path):
    df = pd.read_csv(csv_path)

    # drop rows where prediction is missing
    df = df.dropna(subset=["prediction"])

    # make sure types are ints
    df["label"] = df["label"].astype(int)
    df["prediction"] = df["prediction"].astype(int)

    # confusion matrix values
    tp = len(df[(df["label"] == 1) & (df["prediction"] == 1)])
    tn = len(df[(df["label"] == 0) & (df["prediction"] == 0)])
    fp = len(df[(df["label"] == 0) & (df["prediction"] == 1)])
    fn = len(df[(df["label"] == 1) & (df["prediction"] == 0)])

    # accuracy, precision, recall, F1 all rounded to 4 decimals
    accuracy = (tp + tn) / len(df)
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0
    recall = tp / (tp + fn) if (tp + fn) > 0 else 0
    specificity = tn / (tn + fp) if (tn + fp) > 0 else 0
    f1 = 2 * precision * recall / (precision + recall) if (precision + recall) > 0 else 0

    

    print(f"\nResults for {csv_path}")
    print("Confusion Matrix:")
    print(f"TN: {tn}  FP: {fp}")
    print(f"FN: {fn}  TP: {tp}")
    print(f"Accuracy: {accuracy:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall: {recall:.4f}")
    print(f"Specificity: {specificity:.4f}")
    print(f"F1: {f1:.4f}")

    # false positives and false negatives
    false_positives = df[(df["label"] == 0) & (df["prediction"] == 1)]
    false_negatives = df[(df["label"] == 1) & (df["prediction"] == 0)]

    print("\n5 Random False Positives:")
    if len(false_positives) > 0:
        sample_fp = false_positives.sample(min(5, len(false_positives)), random_state=42)
        for _, row in sample_fp.iterrows():
            print(f"\nIndex: {row['i']}")
            print(f"Text: {row['text']}")
    else:
        print("None")

    print("\n5 Random False Negatives:")
    if len(false_negatives) > 0:
        sample_fn = false_negatives.sample(min(5, len(false_negatives)), random_state=42)
        for _, row in sample_fn.iterrows():
            print(f"\nIndex: {row['i']}")
            print(f"Text: {row['text']}")
    else:
        print("None")


print('Base LLM results')
parse_results('llm_results.csv')


print('\nFew-shot LLM results')
parse_results('few-shot_llm_results.csv')
```
